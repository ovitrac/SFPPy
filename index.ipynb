### Minimal setup for SFPPy core 

> This setup is intended to work on a machine with SFPPy installed or not.

```python
import sys, os
import ipywidgets as widgets
from IPython.display import display, HTML
verbose = True  # Define verbosity flag
try:
    import patankar.loadpubchem
    if verbose: print("SFPPy is already initialized; skipping initialization.")
except ImportError:
    if verbose: print("SFPPy not detected; attempting initialization.")
    base_dirs = ["./SFPPy/utils","../utils", "./utils", ".", "../../utils", "../../../utils"]
    candidates = [os.path.abspath(os.path.join(d, "initialize_sfppy.py")) for d in base_dirs if os.path.exists(os.path.join(d, "initialize_sfppy.py"))]
    if candidates:
        if verbose: print(f"Running initialization script: {candidates[0]}")
        %run -i {candidates[0]} --no-verbose --no-cd
        try:
            import patankar.loadpubchem
            if verbose: print("SFPPy successfully initialized.")
        except ImportError:
            print(f"ERROR: Initialization failed; 'patankar.loadpubchem' is still unavailable.\n","Current sys.path:\n",sys.path)
    else:
        if not os.path.exists("SFPPy"):
            if verbose: print("We clone the repository be patient.")
            !git clone https://github.com/ovitrac/SFPPy.git # First time: clone the repository
        else:
            if verbose: print("We refresh the repository.")
            !git -C SFPPy pull  # Subsequent times: update the repository (if needed)
        %cd SFPPy
        installationscript = os.path.abspath(os.path.join(os.getcwd(),"utils", "initialize_sfppy.py"))
        %run -i {installationscript} --no-verbose --no-cd
        try:
            import patankar.loadpubchem
            if verbose: print("SFPPy successfully initialized.")
        except ImportError:
            print(f"ERROR: Initialization failed; 'patankar.loadpubchem' is still unavailable.\n","Current sys.path:\n",sys.path)
SFPPymainfolder = os.path.abspath(os.path.join(os.path.dirname(patankar.loadpubchem.__file__),".."))
print("SFPPy installation folder:",SFPPymainfolder)
```

### List of Available Notebooks

```python
# Define the subfolder where notebooks are stored (if any)
notebooks_folder = "notebooks"  # If notebooks are stored in a subfolder; otherwise use the main folder

# Determine the directory where notebooks are located:
notebooks_dir = os.path.join(SFPPymainfolder, notebooks_folder) if os.path.exists(os.path.join(SFPPymainfolder, notebooks_folder)) else SFPPymainfolder

# List available notebooks (sorted in ascending order), excluding index.ipynb
notebooks_files = sorted([f for f in os.listdir(notebooks_dir) if f.endswith('.ipynb') and f.lower() != 'index.ipynb'])

# Create a dropdown widget with the notebook filenames
dropdown = widgets.Dropdown(
    options=notebooks_files,
    description="Notebook:",
)

# Function to open the selected notebook locally (relative link)
def open_local_notebook(btn):
    selected_notebook = dropdown.value
    if os.path.exists(os.path.join(SFPPymainfolder, notebooks_folder)):
        # Use a relative path if notebooks are in a subfolder
        local_path = f"{notebooks_folder}/{selected_notebook}"
    else:
        local_path = selected_notebook
    display(HTML(f'<a href="{local_path}" target="_blank">Click here to open {selected_notebook} locally</a>'))

# Function to open the selected notebook remotely on Colab (via GitHub URL)
def open_remote_notebook(btn):
    selected_notebook = dropdown.value
    if os.path.exists(os.path.join(SFPPymainfolder, notebooks_folder)):
        remote_url = f"https://colab.research.google.com/github/ovitrac/SFPPy/blob/main/{notebooks_folder}/{selected_notebook}"
    else:
        remote_url = f"https://colab.research.google.com/github/ovitrac/SFPPy/blob/main/{selected_notebook}"
    display(HTML(f'<a href="{remote_url}" target="_blank">Click here to open {selected_notebook} on Colab</a>'))

# Create two buttons for the two actions
button_local = widgets.Button(description="Open Local Notebook")
button_local.on_click(open_local_notebook)

button_remote = widgets.Button(description="Open Colab Notebook")
button_remote.on_click(open_remote_notebook)

# Display the dropdown and the two buttons
display(dropdown, button_local, button_remote)
```
